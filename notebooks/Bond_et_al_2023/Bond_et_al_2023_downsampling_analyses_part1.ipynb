### Merge loops called from each individual library

```python
import bioframe
import pandas as pd
import numpy as np

def merge_dots(df):
    overlapped_anchors_left=bioframe.cluster(df[['chrom1','start1','end1']].copy().rename(columns={'chrom1': 'chrom', 'start1': 'start', 'end1': 'end'}), min_dist=None)
    overlapped_anchors_right=bioframe.cluster(df[['chrom2','start2','end2','sample_name']].copy().rename(columns={'chrom2': 'chrom', 'start2': 'start', 'end2': 'end'}), min_dist=None)
    overlapped_dots=pd.merge(overlapped_anchors_left, overlapped_anchors_right, left_index=True, right_index=True, suffixes=(1,2))
    sample_name=overlapped_dots.groupby(['cluster1','cluster2'])['sample_name'].apply(lambda x: '&'.join(list(np.unique(x)))).reset_index()
    overlapped_dots=pd.merge(overlapped_dots[['cluster1','cluster2','chrom1','cluster_start1','cluster_end1','chrom2','cluster_start2','cluster_end2']], sample_name, on=['cluster1','cluster2'])
    overlapped_dots=overlapped_dots.drop(['cluster1', 'cluster2'], axis=1).rename(columns={'cluster_start1': 'start1', 'cluster_end1': 'end1', 'cluster_start2': 'start2', 'cluster_end2': 'end2'})
    overlapped_dots=overlapped_dots.drop_duplicates().reset_index(drop=True)
    return overlapped_dots
```

```python
# Bond et al 2023 (old method)

library(ComplexHeatmap)

create_combination_matrix_and_upset_plot <- function(df, top_n=7){
  # Create the list with the desired order
  lt <- setNames(list(
    K562_0000min = which(grepl("K562_0000min", df$sample_name)),
    K562_0360min = which(grepl("K562_0360min", df$sample_name)),
    K562_4320min = which(grepl("K562_4320min", df$sample_name))
  ), c("0 hour", "6 hour", "72 hour")
  )
  
  # Create combination matrix
  m <- make_comb_mat(lt)
  
  # Get the sizes of all combinations
  sizes <- comb_size(m)
  
  # Get the indices of the top n most frequent intersections
  top_n_indices <- order(sizes, decreasing = TRUE)[1:top_n]
  
  # Subset the combination matrix to only those top n
  m_top_n <- m[top_n_indices]
  
  # Set the order of sets (used in plotting, not subsetting)
  set_name_order <- c("0 hour", "6 hour", "72 hour")
  
  upset_plot <- UpSet(
    m_top_n,
    comb_order = seq_along(top_n_indices),  # order is already limited
    set_order = set_name_order,
    top_annotation = upset_top_annotation(m_top_n,
                                          bar_width = 0.5,
                                          add_numbers = TRUE,
                                          numbers_rot = 0, 
                                          numbers_gp = gpar(fontsize = 8),
                                          annotation_name_gp = gpar(fontsize = 10)
    ),
    right_annotation = upset_right_annotation(
      m_top_n,
      bar_width = 0.5,
      add_numbers = TRUE,
      numbers_gp = gpar(fontsize = 8),
      gp = gpar(fill = "#A9A9A9"),
      annotation_name_side = "top",
      axis_param = list(side = "top"),
      annotation_name_gp = gpar(fontsize = 10)
    ),
    row_names_rot = 0,
    row_names_gp = gpar(fontsize = 10),
    pt_size = unit(5, "mm")
  )
  
  return(upset_plot)
}


df_2500M_old <- read.table(file = 'union_dot_list_of_Bond_et_al_2023_5kb_old_method.txt', sep = '\t', header = TRUE)
df_2000M_old <- read.table(file = 'union_dot_list_of_Bond_et_al_2023_5kb_old_method_cis_2000M.txt', sep = '\t', header = TRUE)
df_1500M_old <- read.table(file = 'union_dot_list_of_Bond_et_al_2023_5kb_old_method_cis_1500M.txt', sep = '\t', header = TRUE)
df_1000M_old <- read.table(file = 'union_dot_list_of_Bond_et_al_2023_5kb_old_method_cis_1000M.txt', sep = '\t', header = TRUE)
df_500M_old <- read.table(file = 'union_dot_list_of_Bond_et_al_2023_5kb_old_method_cis_500M.txt', sep = '\t', header = TRUE)
df_500M_10kb_old <- read.table(file = 'union_dot_list_of_Bond_et_al_2023_10kb_old_method_cis_500M.txt', sep = '\t', header = TRUE)

upset_2500M_old <- create_combination_matrix_and_upset_plot(df_2500M_old)
upset_2000M_old <- create_combination_matrix_and_upset_plot(df_2000M_old)
upset_1500M_old <- create_combination_matrix_and_upset_plot(df_1500M_old)
upset_1000M_old <- create_combination_matrix_and_upset_plot(df_1000M_old)
upset_500M_old <- create_combination_matrix_and_upset_plot(df_500M_old)
upset_500M_10kb_old <- create_combination_matrix_and_upset_plot(df_500M_10kb_old)

# Save as PDF
# Save pdf, open pdf in illustrator, it will become svg automatically.
# I uploaded this figure from my laptop to 
# /home/jiangyuan.liu5-umw/projects/4DNII/notebooks/dots_R1R2_updated/Validate_union_list_loop_caller_using_public_data/Bond_et_al_2023/Figures/Figure_7
pdf("upset_plot_of_union_dot_list_of_downsampling_Bond_et_al_2023_5kb_old_method.pdf", width = 4, height = 12)
upset_2500M_old%v%upset_2000M_old%v%upset_1500M_old%v%upset_1000M_old%v%upset_500M_old%v%upset_500M_10kb_old
dev.off()

```

```python
# Bond et al 2023 (new method)

library(ComplexHeatmap)

create_combination_matrix_and_upset_plot <- function(df, top_n=7){
  # Create the list with the desired order
  lt <- setNames(list(
    K562_0000min = which(grepl("K562_0000min", df$sample_name)),
    K562_0360min = which(grepl("K562_0360min", df$sample_name)),
    K562_4320min = which(grepl("K562_4320min", df$sample_name))
  ), c("0 hour", "6 hour", "72 hour")
  )
  
  # Create combination matrix
  m <- make_comb_mat(lt)
  
  # Get the sizes of all combinations
  sizes <- comb_size(m)
  
  # Get the indices of the top n most frequent intersections
  top_n_indices <- order(sizes, decreasing = TRUE)[1:top_n]
  
  # Subset the combination matrix to only those top n
  m_top_n <- m[top_n_indices]
  
  # Set the order of sets (used in plotting, not subsetting)
  set_name_order <- c("0 hour", "6 hour", "72 hour")
  
  upset_plot <- UpSet(
    m_top_n,
    comb_order = seq_along(top_n_indices),  # order is already limited
    set_order = set_name_order,
    top_annotation = upset_top_annotation(m_top_n,
                                          bar_width = 0.5,
                                          add_numbers = TRUE,
                                          numbers_rot = 0, 
                                          numbers_gp = gpar(fontsize = 8),
                                          annotation_name_gp = gpar(fontsize = 10)
    ),
    right_annotation = upset_right_annotation(
      m_top_n,
      bar_width = 0.5,
      add_numbers = TRUE,
      numbers_gp = gpar(fontsize = 8),
      gp = gpar(fill = "#A9A9A9"),
      annotation_name_side = "top",
      axis_param = list(side = "top"),
      annotation_name_gp = gpar(fontsize = 10)
    ),
    row_names_rot = 0,
    row_names_gp = gpar(fontsize = 10),
    pt_size = unit(5, "mm")
  )
  
  return(upset_plot)
}

df_2500M_new <- read.table(file = 'Bond_et_al_2023_union_loop_list_5kb.tsv', sep = '\t', header = TRUE)
df_2000M_new <- read.table(file = 'Bond_et_al_2023_union_loop_list_5kb_cis_2000M.tsv', sep = '\t', header = TRUE)
df_1500M_new <- read.table(file = 'Bond_et_al_2023_union_loop_list_5kb_cis_1500M.tsv', sep = '\t', header = TRUE)
df_1000M_new <- read.table(file = 'Bond_et_al_2023_union_loop_list_5kb_cis_1000M.tsv', sep = '\t', header = TRUE)
df_500M_new <- read.table(file = 'Bond_et_al_2023_union_loop_list_5kb_cis_500M.tsv', sep = '\t', header = TRUE)
df_500M_10kb_new <- read.table(file = 'Bond_et_al_2023_union_loop_list_10kb_cis_500M.tsv', sep = '\t', header = TRUE)

upset_2500M_new <- create_combination_matrix_and_upset_plot(df_2500M_new)
upset_2000M_new <- create_combination_matrix_and_upset_plot(df_2000M_new)
upset_1500M_new <- create_combination_matrix_and_upset_plot(df_1500M_new)
upset_1000M_new <- create_combination_matrix_and_upset_plot(df_1000M_new)
upset_500M_new <- create_combination_matrix_and_upset_plot(df_500M_new)
upset_500M_10kb_new <- create_combination_matrix_and_upset_plot(df_500M_10kb_new)

# Save as PDF
# Save pdf, open pdf in illustrator, it will become svg automatically.
# I uploaded this figure from my laptop to 
# /home/jiangyuan.liu5-umw/projects/4DNII/notebooks/dots_R1R2_updated/Validate_union_list_loop_caller_using_public_data/Bond_et_al_2023/Figures/Figure_7
pdf("upset_plot_of_union_dot_list_of_downsampling_Bond_et_al_2023_5kb_new_method.pdf", width = 4, height = 12)
upset_2500M_new%v%upset_2000M_new%v%upset_1500M_new%v%upset_1000M_new%v%upset_500M_new%v%upset_500M_10kb_new
dev.off()
```
